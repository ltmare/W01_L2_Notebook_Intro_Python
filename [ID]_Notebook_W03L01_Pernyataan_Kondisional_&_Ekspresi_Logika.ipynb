## 3. ELABORATE: AKT-A02 · Program Ongkos Kirim
Selesaikan permasalahan penentuan ongkos kirim berdasarkan spesifikasi aturan berikut:

### Aturan Ongkos Kirim:
* Berat $\le 1$ kg $\rightarrow$ Rp 10.000
* $1$ kg $<$ Berat $\le 5$ kg $\rightarrow$ Rp 25.000
* Berat $> 5$ kg $\rightarrow$ Rp 25.000 + Rp 4.000 per kg kelebihan beratnya.
* Jika pembeli adalah **Member** DAN total ongkos $\ge$ Rp 30.000, maka akan mendapatkan **potongan harga sebesar 10%**.

```python
def hitung_ongkir(berat, is_member):
    # 1. Hitung Ongkos Kirim Dasar berdasarkan Berat
    if berat <= 1:
        ongkos = 10000
    elif berat <= 5:
        ongkos = 25000
    else:
        kelebihan = berat - 5
        ongkos = 25000 + (kelebihan * 4000)

    # 2. Terapkan Potongan Member (Diskon 10% jika ongkos >= Rp30.000)
    if is_member and ongkos >= 30000:
        diskon = 0.10 * ongkos
        ongkos = ongkos - diskon
        print(f"Selamat! Anda mendapatkan potongan member sebesar Rp {int(diskon):,}")

    return int(ongkos)

# --- Uji Coba Kasus ---
# Uji beberapa input seperti: 0.5 kg, 1 kg, 5 kg, dan 7.5 kg (baik Member maupun Non-Member)
berat_input = float(input("Masukkan berat paket (kg): "))
member_status = input("Apakah member? (y/n): ").lower() == 'y'

total_bayar = hitung_ongkir(berat_input, member_status)
print(f"Total Ongkos Kirim yang harus dibayar: Rp {total_bayar:,}")
```
