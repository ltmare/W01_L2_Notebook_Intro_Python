## 2.1 Urutan Operator Aritmatika (Operator Precedence)

Python mengikuti aturan urutan operasi matematika standar (PEMDAS/BODMAS). Ini berarti operator tertentu memiliki prioritas lebih tinggi daripada yang lain.

````python
# Contoh urutan operasi

# Perkalian dan pembagian memiliki prioritas lebih tinggi daripada penjumlahan dan pengurangan
print("2 + 3 * 4 =", 2 + 3 * 4) # Seharusnya (3 * 4) dulu, lalu + 2


# Menggunakan tanda kurung untuk mengubah urutan
print("(2 + 3) * 4 =", (2 + 3) * 4) # Seharusnya (2 + 3) dulu, lalu * 4

# Eksponen memiliki prioritas tertinggi
print("2 ** 3 + 1 =", 2 ** 3 + 1) # Seharusnya (2 ** 3) dulu, lalu + 1

# Kombinasi
print("10 - 2 * 3 / 2 =", 10 - 2 * 3 / 2) # (2 * 3) = 6, lalu (6 / 2) = 3, lalu 10 - 3 = 7.0
````

## 3. Operasi arithmatic dengan string

Dalam Python, Anda dapat melakukan operasi seperti penyambungan (concatenation) dan pengulangan (repetition) pada string menggunakan operator aritmetika `+` dan `*`.

````python
# Penyambungan (Concatenation) string menggunakan operator '+'
print("Hello " + "World!")
print("Python " + "adalah " + "menyenangkan!")
````

````python
# Pengulangan (Repetition) string menggunakan operator '*'
print("Na " * 3)
print("=" * 20)
print("Ini contoh " + ("baris " * 2))
````

## 5. Karakter Khusus (Special Characters)

Karakter khusus digunakan dalam string untuk melakukan tindakan tertentu, seperti membuat baris baru atau tab. Mereka diawali dengan backslash (`\`).

````python
# Karakter Newline (\n) untuk membuat baris baru
print("Baris pertama.\nBaris kedua.")

# Karakter Tab (\t) untuk membuat indentasi
print("Kolom 1\tKolom 2\tKolom 3")
print("Data 1\tData 2\tData 3")

# Menggabungkan keduanya
print("Nama:\tBudi\nUsia:\t25")
````
