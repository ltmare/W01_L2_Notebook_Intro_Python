## Aktivitas 1: Membaca Input dan Menggunakan F-String
Fungsi `input()` berfungsi untuk menunggu ketikan pengguna dan membaca masukannya. Nilai yang dikembalikan oleh `input()` akan selalu berupa teks (`str`), meskipun pengguna mengetikkan angka.

Untuk menyisipkan nilai variabel langsung ke dalam sebuah kalimat teks, kita dapat menggunakan **f-string** dengan meletakkan huruf `f` di depan tanda kutip dan menyisipkan variabel di dalam tanda kurung kurawal `{}`.

**Tugas:** Jalankan kode di bawah ini dan cobalah memasukkan nama kota Anda.


```python
# Menampilkan pesan dan meminta masukan pengguna
kota = input("Nama kota asalmu? ")

# Cetak kalimat sapaan menggunakan f-string
# Sisipan di dalam kurung kurawal {} akan diganti dengan nilainya[span_8](start_span)[span_8](end_span)
print(f"Salam dari {kota}!")
```

## Aktivitas 2: Casting (Konversi Tipe Data)
Empat tipe data dasar di Python adalah bilangan bulat (`int`), bilangan desimal (`float`), teks (`str`), dan logika benar/salah (`bool`)[.

Ketika kita membaca angka menggunakan `input()`, nilai tersebut masih berupa string (teks). Jika teks dikalikan dengan angka (misal `"3" * 2`), maka teks tersebut hanya akan diulang menjadi `"33"`, bukan dikalikan secara matematika. Jika dijumlahkan (`str + int`), program akan berhenti dan memunculkan `TypeError`. Oleh karena itu, kita butuh melakukan **casting** atau konversi tipe data.

**Tugas:** Perbaiki program perhitungan tiket di bawah ini dengan melakukan casting pada variabel `jumlah_teks` menjadi tipe `int` (bilangan bulat).


```python
jumlah_teks = input("Berapa tiket? ")

# TODO: Gunakan fungsi int() untuk mengubah teks menjadi bilangan bulat.
# Simpan hasil konversinya ke dalam variabel 'jumlah[span_13](start_span)'[span_13](end_span).
jumlah = int(jumlah_teks)

# Menghitung dan mencetak total.
# Isi di dalam {} bisa berupa hitungan matematika yang dihitung terlebih dulu[span_14](start_span)[span_14](end_span).
print(f"Total: {jumlah * 2}")
```

## Aktivitas 3: Operasi Pembagian
Operator bisa menghasilkan tipe data yang berbeda tergantung fungsinya.
* Operator pembagian biasa `/` akan **selalu** menghasilkan bilangan desimal (`float`).
* Operator pembagian lantai `//` (floor division) berfungsi untuk membulatkan ke bawah, bukan sekadar memotong desimalnya.

**Tugas:** Jalankan kode berikut untuk membuktikan hasil operasi tipe data. Gunakan fungsi `type()` saat Anda ragu untuk mengetahui tipe data suatu objek.


```python
hasil_float = 3 / 1
print("Hasil 3 / 1 adalah:", hasil_float)
print("Tipe data dari hasil di atas:", type(hasil_float)) # Menghasilkan <class 'float'>[span_18](start_span)[span_18](end_span)

hasil_floor = -7 // 2
print("\nHasil pembagian lantai -7 // 2 adalah:", hasil_floor) # Dibulatkan ke bawah menjadi -4[span_19](start_span)[span_19](end_span)
print("Tipe data dari hasil di atas:", type(hasil_floor))
```

## Aktivitas 4: Klinik Bug dan Rancang Bangun
Saat melakukan konversi (casting), ingatlah bahwa `int()` untuk bilangan bulat dan `float()` untuk nilai apa pun yang mungkin bernilai desimal. Fungsi `int()` akan memotong angka desimal ke arah nol (tidak membulatkan), dan ia tidak bisa membaca nilai string yang mengandung titik desimal, misalnya `int("12.5")` akan gagal/error.

**Tugas:** Lengkapilah rancangan program perhitungan subtotal pembelian. Aturan penamaan yang baik adalah: **satu nama, satu makna, untuk seluruh program**.


```python
# 1. Baca harga satuan dan jumlah beli sebagai teks[span_23](start_span)[span_23](end_span)
harga_teks = input("Harga satuan: ")
jumlah_teks = input("Jumlah beli: ")

# 2. Ubah tipe (Casting)[span_24](start_span)[span_24](end_span)
# TODO: Ubah harga_teks menjadi float dan simpan di variabel 'harga'
harga = float(harga_teks)

# TODO: Ubah jumlah_teks menjadi int dan simpan di variabel 'jumlah'
jumlah = int(jumlah_teks)

# 3. Hitung subtotal[span_25](start_span)[span_25](end_span)
subtotal = harga * jumlah

# 4. Tampilkan subtotal[span_26](start_span)[span_26](end_span)
print(f"Subtotal pembelian: {subtotal}")
```
